```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# Part 1: Data **Generation**

```python
# Uniform distribution (Regular days)
uniform_wait = np.random.uniform(2, 12, 3000)

# Normal distribution (Busy days)
normal_wait = np.random.normal(8, 2, 3000)

# Replace negative values with 0
normal_wait[normal_wait < 0] = 0
```

# Part 2: Visualization

```python
# Histogram for Uniform Distribution
plt.hist(uniform_wait, bins=30, alpha=0.5, color='blue', edgecolor='black')
plt.title('Uniform Distribution (Regular Days)')
plt.xlabel('Wait Time')
plt.ylabel('Frequency')
plt.show()

# Histogram for Normal Distribution
plt.hist(normal_wait, bins=30, alpha=0.5, color='green', edgecolor='black')
plt.title('Normal Distribution (Busy Days)')
plt.xlabel('Wait Time')
plt.ylabel('Frequency')
plt.show()
```

# Part 3: Statistical **Analysis**

```python
print("Regular Days (Uniform)")
print("Mean:", np.mean(uniform_wait))
print("Median:", np.median(uniform_wait))
print("Standard Deviation:", np.std(uniform_wait))
print("Max:", np.max(uniform_wait))

print("\nBusy Days (Normal)")
print("Mean:", np.mean(normal_wait))
print("Median:", np.median(normal_wait))
print("Standard Deviation:", np.std(normal_wait))
print("Max:", np.max(normal_wait))
```
